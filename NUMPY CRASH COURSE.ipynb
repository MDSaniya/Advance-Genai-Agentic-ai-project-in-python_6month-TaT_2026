# 2. Array Manipulation Functions

```python
# Reshape an array
a1 = np.array([1,2,3])
reshaped = np.reshape(a1,(1,3)) # Reshape to 1x3
print("Reshape array:",reshaped)
```

```python
a1 = np.array([1,2,3])
reshaped = np.reshape(a1,(3,1)) #Reshape to 3x1
print("Reshaped array:",reshaped)
```

```python
# Flatten an array
f1 = np.array([[1,2],[3,4],[5,6],[7,8]])
flattened = np.ravel(f1) #flatten to 1D array
print("Flattened array:",flattened)
```

```python
# Transposed an array
e1 = np.array([[1,2],[3,4]])
transposed = np.transpose(e1) # Transpose the array
print("Transposed Array:\n", transposed) # it will change rows into columns and columns into rows
```

```python
# stack arrays vertically
a2 = np.array([1,2])
b2 = np.array([3,4])
stacked = np.vstack([a2,b2]) # stack a and b vertically
print("Stacked arrays:\n",stacked)
```

```python
# stack arrays horizontally
a2 = np.array([1,2])
b2 = np.array([3,4])
stacked = np.hstack([a2,b2]) # stack a and b horizontally
print("Stacked arrays:\n",stacked)
```

# 4. Statistical Functions

```python
s = np.array([1,2,3,4])
mean = np.mean(s)
print("Mean of s",mean) # sum of numbers / total numbers
```

```python
# Variance of an array
var = np.var(s) # (1-2.5)^2 + (2-2.5)^2 + (3-2.5)^2 + (4-2.5)^2 / 4
print("variance of s:",var)
```

```python
# Standard deviation of an array
std_dev = np.std(s)
print("Standard deviation of s:",std_dev) # root of variance will give std val in float
```

```python
# Minimum ele of an array
minimum = np.min(s)
print("Min of s:",minimum)
```

```python
# maximum ele of an array
maximum = np.max(s)
print("Min of s:",maximum)
```

# 6.Boolean and Logical functions

```python
#check if all elements are true
#all
logical_test = np.array([True,False,True])
all_true = np.all(logical_test) # check if all are true
print("All elements True:",all_true)



```

```python
#Check if all elements are true
logical_test = np.array([False,False,False])
all_true = np.all(logical_test) # check if all are true
print("All elements True:",all_true)
```

```python
#check if any element are true
#any
any_true = np.any(logical_test) #check if any are true
print("Any elements True:",any_true)
```

# 7. Set Operations

```python
#Intersection of two arrays
set_a = np.array([1,2,3,4])
set_b = np.array([3,4,5,6])
intersection = np.intersect1d(set_a,set_b)
print("Intersection of a and b:",intersection)
```

```python
#Union of two arrays
union = np.union1d(set_a,set_b)
print("Union of a and b:",union)
```

# 8.Array Attribute Functions

```python
#Array attributes
a = np.array([1,2,3])
shape = a.shape #shape of the array
size = a.size # number of ele
dimensions = a.ndim # num of dimensions
dtype = a.dtype #data type of te array

print("Sahpe of a:",shape)
print("Size of a:",size)
print("Number of dimensions of a:",dimensions)
print("Data type of a:",dtype)
```
