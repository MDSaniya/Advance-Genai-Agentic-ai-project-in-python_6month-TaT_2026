# Seaborn

```python
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
plt.rcParams['figure.figsize']=6,8 #size of the figure

import warnings
warnings.filterwarnings('ignore') # os error
```

```python
df['InternetUsers']
```

```python
vis1 = sns.distplot(df["InternetUsers"]) # univariant analysis using only one variable,distplot() was an older Seaborn function used to plot a histogram along with a KDE curve.
```

```python
vis1 = sns.displot(df["InternetUsers"]) #displot() is the modern figure-level function for displaying distributions.
```

```python
vis1 = sns.distplot(df["InternetUsers"],bins=15) # it will create 15 bins
```

# BOX PLOTS:

```python
vis4 = sns.boxplot(data = df, x = "IncomeGroup",y="BirthRate") # Bi-Variate analysis
```

# LINEAR MODEL PLOT

```python
vis5 = sns.lmplot(data = df, x = "InternetUsers",y="BirthRate") #linear model plot
```

```python
vis6 = sns.lmplot(data = df, x = "InternetUsers",y="BirthRate",fit_reg = False) #no regression line
```

```python
vis7 = sns.lmplot(data = df, x = "InternetUsers",y="BirthRate",fit_reg = False,hue = "IncomeGroup") # hue - parameter for color
```

```python
vis7 = sns.lmplot(data = df, x = "InternetUsers",y="BirthRate",fit_reg = True,hue = "IncomeGroup")
```
